```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time

# -------- CONFIG --------
CHAT_URL = "https://3.29.174.173/"   # <--- UPDATE THIS


# -------- SETUP --------
options = Options()
driver = webdriver.Chrome(options=options)

# -------- OPEN PAGE --------
driver.get(CHAT_URL)

wait = WebDriverWait(driver, 30000)

# -------- SEND QUESTION --------
input_box = wait.until(EC.presence_of_element_located((By.ID, "messageInput")))
send_btn  = wait.until(EC.presence_of_element_located((By.ID, "sendButton")))

```

```python


import pandas as pd

df = pd.read_csv("C:/Users/maryam.maksour/OneDrive - AL BAYARI/Desktop/Book1_with_responses.csv")
 

```

```python
df["Response"] = pd.Series(index=df.index, dtype="string")
df["SQL Statement"] = pd.Series(index=df.index, dtype="string")
```

```python
print(df['Question'][7])
```

```python
i = 0
for i in range(len(df["Question"])):

	if not df['Response'][i] == "Error: Could not connect to the server":
		continue
	 
	question = df["Question"][i]

	input_box.clear()
	input_box.send_keys(question)
	send_btn.click()


	# -------- GET REPLY --------
	reply = None
	timeout = time.time() + 3000   

	while time.time() < timeout:
		messages = driver.find_elements(By.CSS_SELECTOR,
			"div.message.bot, div.message.assistant, div[class*='message']"
		)
		if messages:
			text = messages[-1].text.strip()
			if text:
				reply = text
				break
		time.sleep(1)
		
		# -------- PRINT REPLY --------
	res = reply.split('Source: db')
	df.loc[i, "Response"] = str(res[0])
	df.loc[i, "SQL Statement"] = str(res[1]) if len(res) > 1 else ""

	time.sleep(3)
 

```
